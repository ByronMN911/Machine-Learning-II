```python
import pandas as pd
from sklearn.linear_model import LinearRegression
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Creamos el conjunto de datos
df = pd.DataFrame({
    "horas":[1,2,3,4,5],
    "nota": [2,3,5,4,6]
})

X= df[["horas"]]
y= df[["nota"]]
print (X)

```

```python
#Creamos nuestro modeo
modelo = LinearRegression()
modelo.fit(X, y)
```

```python
#Imprimimos el interceptor y el coef
print("Inteceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
#Valores predichos
y_pred = modelo.predict(X)
```

```python
#Realizamos la nueva predicción 
horas = float(input("ingrese las horas de estudio"))
prediccion = modelo.predict([[horas]])
print("predicción:", prediccion)
```

```python
#Graficar con matplotlib el modelo y verificar si el nuevo valor corresponde a la predicción en la recta
plt.figure(figsize=(15,5)) #Este solo es el lienzo donde irá la figura con un ancho de 15 inchs y un alto de 5 inch
X_grafica = np.append(X, horas)
y_grafica = np.append(y_pred, prediccion)
plt.scatter(X, y)
plt.xlabel("Horas de Estudio")
plt.ylabel("Notas")
plt.title("Regresión Lineal - Relación horas de estudio con notas")

plt.plot(X_grafica, y_grafica)
plt.scatter(X, y_pred, c="orange")
plt.scatter(horas, prediccion, c="green")

plt.show()
```
